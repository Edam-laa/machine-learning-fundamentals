# Logistic Regression for Admissions

**Objective:** Model the probability of student admission from academic features using logistic regression.

**Dataset used:** `data/admissions.csv`

**Main techniques used:** Data visualization, sigmoid function, logistic regression, probability interpretation, decision thresholding


```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the admissions data
admissions = pd.read_csv('data/admissions.csv')

# Create scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(admissions['gpa'], admissions['admit'], alpha=0.7)

# Add labels and title
plt.xlabel('GPA')
plt.ylabel('Admission Status')
plt.title('Relationship between GPA and Admission Status')

# Display the plot
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
#fonction logestic
def logistic(x):
    #np.exp() calcul l'exp
    return np.exp(x)/ (1+np.exp(x))
#generer 50 valeur reelles entre -6 et 6
x = np.linspace(-6,6,50, dtype=float)
#transformer les nombre via fonct
y=logistic(x)
#tracer la courbe
plt.plot(x,y)
plt.ylabel("probabilité")
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(solver='liblinear')
logistic_model.fit(admissions[['gpa']], admissions['admit'])
```

```python
#calcl du probabilite avec un parametre c'est une matrice d'observation
probabilities = logistic_model.predict_proba(admissions[['gpa']])
```

```python
#probabilite que la ligne apprt au label 0
probabilities[:,0]
```

```python
#probabilite que la ligne apprt au label 1
probabilities[:,1]
```
